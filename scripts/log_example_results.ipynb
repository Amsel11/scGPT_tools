```python
import scanpy as sc
import matplotlib.pyplot as plt

# Load the results
results = sc.read_h5ad('/root/scGPT_dir/scGPT/save/embeddings_20250409_10/embeddings_pred_results_20250409_234208.h5ad')

# Plot cell type distribution
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
results.obs['cell_type'].value_counts().plot(kind='bar', ax=ax1)
ax1.set_title('True Cell Type Distribution')
results.obs['pred_cell_type'].value_counts().plot(kind='bar', ax=ax2)
ax2.set_title('Predicted Cell Type Distribution')
plt.tight_layout()
plt.show()

# Create a confusion matrix
from sklearn.metrics import confusion_matrix
import seaborn as sns

conf_mat = confusion_matrix(
    results.obs['cell_type'], 
    results.obs['pred_cell_type'],
    labels=results.obs['cell_type'].unique()
)

plt.figure(figsize=(10, 8))
sns.heatmap(conf_mat, annot=True, fmt='d', 
            xticklabels=results.obs['cell_type'].unique(),
            yticklabels=results.obs['cell_type'].unique())
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.show()



```
